```python
def view_gdb_layers():
    import fiona
    import geopandas as gpd
    import tkinter as tk
    from tkinter import messagebox

    # Ask user for GDB path in the interpreter (no Tkinter prompt)
    gdb_path = input("Enter the full path to your geodatabase (.gdb): ").strip()

    if not gdb_path:
        print("No path provided.")
        return

    # Try reading layers
    try:
        layers = fiona.listlayers(gdb_path)
        print("\nGeodatabase loaded successfully.")
    except Exception as e:
        print(f"\nError reading GDB:\n{e}")
        return

    # Tkinter window for selecting layers
    root = tk.Tk()
    root.title("Geodatabase Layer Viewer")
    root.geometry("350x400")

    tk.Label(root, text="Select a Layer:", font=("Arial", 12, "bold")).pack(pady=5)

    listbox = tk.Listbox(root, width=40, height=15)
    listbox.pack(pady=5)

    # Insert all layers into the listbox
    for layer in layers:
        listbox.insert(tk.END, layer)

    def show_columns():
        selected = listbox.curselection()
        if not selected:
            messagebox.showwarning("No Selection", "Please select a layer.")
            return
        
        layer_name = layers[selected[0]]

        try:
            gdf = gpd.read_file(gdb_path, layer=layer_name)
            columns = gdf.columns.tolist()

            # Horizontal display
            horizontal = " | ".join(columns)

            messagebox.showinfo(f"Columns in {layer_name}", horizontal)
        except Exception as e:
            messagebox.showerror("Error", str(e))

    # Button to show columns
    tk.Button(root, text="Show Columns", command=show_columns).pack(pady=10)

    root.mainloop()
```

```python
view_gdb_layers()
```
