```python
import arcpy
import os

def add_prefix_to_feature_classes(gdb_path, prefix):
    """
    Adds a prefix to all feature classes in a geodatabase.
    
    Parameters:
    gdb_path (str): Path to the geodatabase
    prefix (str): Prefix to add to feature class names
    """
    
    # Set workspace to the geodatabase
    arcpy.env.workspace = gdb_path
    
    # Get list of all feature classes in the geodatabase
    feature_classes = arcpy.ListFeatureClasses()
    
    if not feature_classes:
        print("No feature classes found in the geodatabase.")
        return
    
    print(f"\nFound {len(feature_classes)} feature classes to process:")
    
    # Process each feature class
    for fc in feature_classes:
        original_name = fc
        new_name = prefix + original_name
        
        # Skip if the feature class already has the prefix
        if original_name.startswith(prefix):
            print(f"Skipping '{original_name}' - already has prefix")
            continue
            
        try:
            # Rename the feature class
            arcpy.Rename_management(original_name, new_name)
            print(f"Renamed '{original_name}' to '{new_name}'")
        except arcpy.ExecuteError as e:
            print(f"Error renaming '{original_name}': {e}")
    
    print("\nProcessing complete.")

def get_user_input():
    """Gets geodatabase path and prefix from user input"""
    
    print("=== Feature Class Prefix Adder ===")
    
    # Get geodatabase path
    while True:
        gdb_path = input("\nEnter path to geodatabase (.gdb): ").strip()
        if not gdb_path:
            print("Please enter a path.")
            continue
        if not os.path.exists(gdb_path):
            print(f"Path does not exist: {gdb_path}")
            continue
        if not gdb_path.endswith('.gdb'):
            print("Please enter a path to a file geodatabase (.gdb)")
            continue
        break
    
    # Get prefix
    while True:
        prefix = input("\nEnter prefix to add (no spaces or special characters): ").strip()
        if not prefix:
            print("Please enter a prefix.")
            continue
        if not prefix.isidentifier():
            print("Prefix should only contain letters, numbers, and underscores, and cannot start with a number.")
            continue
        break
    
    # Confirm with user
    print(f"\nAbout to add prefix '{prefix}' to all feature classes in: {gdb_path}")
    confirm = input("Continue? (y/n): ").strip().lower()
    
    if confirm != 'y':
        print("Operation cancelled.")
        return None, None
    
    return gdb_path, prefix

# Main execution
if __name__ == "__main__":
    try:
        gdb_path, prefix = get_user_input()
        if gdb_path and prefix:
            add_prefix_to_feature_classes(gdb_path, prefix)
    except Exception as e:
        print(f"\nAn unexpected error occurred: {e}")
    finally:
        input("\nPress Enter to exit...")
```

```python
import arcpy
import os

def remove_prefix_from_feature_classes(gdb_path, prefix):
    """
    Removes a prefix from all feature classes in a geodatabase.
    
    Parameters:
    gdb_path (str): Path to the geodatabase
    prefix (str): Prefix to remove from feature class names
    """
    
    # Set workspace to the geodatabase
    arcpy.env.workspace = gdb_path
    
    # Get list of all feature classes in the geodatabase
    feature_classes = arcpy.ListFeatureClasses()
    
    if not feature_classes:
        print("No feature classes found in the geodatabase.")
        return
    
    print(f"\nFound {len(feature_classes)} feature classes to process:")
    
    # Process each feature class
    for fc in feature_classes:
        original_name = fc
        
        # Skip if the feature class doesn't have the prefix
        if not original_name.startswith(prefix):
            print(f"Skipping '{original_name}' - doesn't start with prefix")
            continue
            
        new_name = original_name[len(prefix):]
        
        # Check if new name would be empty
        if not new_name:
            print(f"Cannot rename '{original_name}' - would result in empty name")
            continue
            
        try:
            # Rename the feature class
            arcpy.Rename_management(original_name, new_name)
            print(f"Renamed '{original_name}' to '{new_name}'")
        except arcpy.ExecuteError as e:
            print(f"Error renaming '{original_name}': {e}")
    
    print("\nProcessing complete.")

def get_user_input():
    """Gets geodatabase path and prefix from user input"""
    
    print("=== Feature Class Prefix Remover ===")
    
    # Get geodatabase path
    while True:
        gdb_path = input("\nEnter path to geodatabase (.gdb): ").strip()
        if not gdb_path:
            print("Please enter a path.")
            continue
        if not os.path.exists(gdb_path):
            print(f"Path does not exist: {gdb_path}")
            continue
        if not gdb_path.endswith('.gdb'):
            print("Please enter a path to a file geodatabase (.gdb)")
            continue
        break
    
    # Get prefix to remove
    while True:
        prefix = input("\nEnter prefix to remove: ").strip()
        if not prefix:
            print("Please enter a prefix.")
            continue
        break
    
    # Confirm with user
    print(f"\nAbout to remove prefix '{prefix}' from feature classes in: {gdb_path}")
    print("WARNING: This operation cannot be undone!")
    confirm = input("Continue? (y/n): ").strip().lower()
    
    if confirm != 'y':
        print("Operation cancelled.")
        return None, None
    
    return gdb_path, prefix

# Main execution
if __name__ == "__main__":
    try:
        gdb_path, prefix = get_user_input()
        if gdb_path and prefix:
            remove_prefix_from_feature_classes(gdb_path, prefix)
    except Exception as e:
        print(f"\nAn unexpected error occurred: {e}")
    finally:
        input("\nPress Enter to exit...")
```
